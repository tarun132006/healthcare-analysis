```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/diabetes_cleaned.csv')
df.head()
```

```python
corr = df.corr()
corr
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title("Correlation Heatmap", fontsize=20)
plt.show()
```

```python
outcome_corr = corr['Outcome'].drop('Outcome')
outcome_corr
```

```python
outcome_corr.sort_values().plot(kind='barh', color='steelblue', figsize=(10, 6))
plt.title("Feature Correlation with Diabetes Outcome", fontsize=20)
plt.xlabel("Correlation Score", fontsize=15)
plt.axvline(x=0, color='black', linewidth=0.5)
plt.show()
```

```python
## Feature Importance
Glucose is the strongest predictor of diabetes (0.49), followed by 
BMI (0.31) and Age (0.24). This ranking confirms our visual findings 
from the distribution charts.
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 6))

df.boxplot(column='Glucose', by='Outcome', ax=axes[0])
df.boxplot(column='BMI', by='Outcome', ax=axes[1])
df.boxplot(column='Age', by='Outcome', ax=axes[2])

plt.suptitle("Top 3 Features by Outcome", fontsize=20)
plt.tight_layout()
plt.show()
```
