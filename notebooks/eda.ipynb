```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/diabetes_cleaned.csv')
df.head()
```

```python
outcome = df["Outcome"].value_counts()

plt.bar(outcome.index, outcome.values, color=['steelblue', 'tomato'],width=0.4)

plt.xlabel("Outcome", fontsize = 20)
plt.ylabel("Patients", fontsize = 20)
plt.title("Diabetes Status", fontsize = 25)

plt.xticks([0, 1], ['Healthy', 'Diabetic'])

plt.show()
```

```python
## Chart 1 - Diabetes Status
The dataset is imbalanced — there are significantly more healthy patients (500) 
than diabetic patients (268). This means roughly 1 in 3 patients in this 
dataset has diabetes.
```

```python
healthy = df[df['Outcome'] == 0]
diabetic = df[df['Outcome'] == 1]

plt.hist(healthy['Glucose'], alpha=0.5, label='Healthy', bins=30)
plt.hist(diabetic['Glucose'], alpha=0.5, label='Diabetic', bins=30)

plt.xlabel("Glucose", fontsize = 20)
plt.ylabel("Patients", fontsize = 20)
plt.title("Glucose Distribution", fontsize = 25)

plt.legend()

plt.show()
```

```python
## Chart 2 - Glucose Distribution by Outcome
Diabetic patients clearly show higher glucose levels than healthy patients.
The two distributions are visibly separated, making Glucose the strongest 
visual predictor of diabetes in this dataset.
```

```python
healthy = df[df['Outcome'] == 0]
diabetic = df[df['Outcome'] == 1]

plt.hist(healthy['Age'], alpha=0.5, label='Healthy', bins=30)
plt.hist(diabetic['Age'], alpha=0.5, label='Diabetic', bins=30)

plt.xlabel("Age of Patients", fontsize = 20)
plt.ylabel("Patients", fontsize = 20)
plt.title("Age Distribution", fontsize = 25)

plt.legend()

plt.show()
```

```python
## Chart 3 - Age Distribution by Outcome
Diabetic patients tend to be older than healthy patients. However the two 
groups overlap more than Glucose did, suggesting age is a contributing 
factor but not as strong a predictor as glucose levels.
```

```python
healthy = df[df['Outcome'] == 0]
diabetic = df[df['Outcome'] == 1]

plt.hist(healthy['BMI'], alpha=0.5, label='Healthy', bins=30)
plt.hist(diabetic['BMI'], alpha=0.5, label='Diabetic', bins=30)

plt.xlabel("BMI of Patients", fontsize = 20)
plt.ylabel("Patients", fontsize = 20)
plt.title("BMI Distribution", fontsize = 25)

plt.legend()

plt.show()
```

```python
## Chart 4 - BMI Distribution by Outcome
Diabetic patients tend to have higher BMI than healthy patients. 
The overlap is significant but the diabetic group clearly skews 
towards higher BMI values.
```

```python
healthy = df[df['Outcome'] == 0]
diabetic = df[df['Outcome'] == 1]

plt.scatter(healthy['Glucose'], healthy['BMI'], alpha=0.5, label='Healthy', color='steelblue')
plt.scatter(diabetic['Glucose'], diabetic['BMI'], alpha=0.5, label='Diabetic', color='tomato')

plt.xlabel("Glucose Levels", fontsize=20)
plt.ylabel("BMI of Patients", fontsize=20)
plt.title("Glucose vs BMI by Outcome", fontsize=25)
plt.legend()
plt.show()
```

```python
## Chart 1 - Glucose vs BMI Scatter Plot
Diabetic patients tend to cluster in the higher glucose and higher BMI 
region. However significant overlap exists between groups, confirming 
that no single pair of features perfectly predicts diabetes alone.
```

```python
healthy = df[df['Outcome'] == 0]
diabetic = df[df['Outcome'] == 1]

plt.scatter(healthy['Glucose'], healthy['Age'], alpha=0.5, label='Healthy', color='steelblue')
plt.scatter(diabetic['Glucose'], diabetic['Age'], alpha=0.5, label='Diabetic', color='tomato')

plt.xlabel("Glucose Levels", fontsize=20)
plt.ylabel("Age of Patients", fontsize=20)
plt.title("Glucose vs Age by Outcome", fontsize=25)
plt.legend()
plt.show()
```

```python
## Chart 2 - Glucose vs Age Scatter Plot
Diabetic patients cluster towards the top right — older age combined 
with higher glucose levels is a strong indicator of diabetes. Younger 
patients with low glucose are almost exclusively healthy (blue dots).
```
